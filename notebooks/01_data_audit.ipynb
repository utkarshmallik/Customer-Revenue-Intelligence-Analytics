```python
import numpy as np

conditions = [
    # 1. Champions
    (rfm["R_score"] >= 4) &
    (rfm["F_score"] >= 4) &
    (rfm["M_score"] >= 4),

    # 2. Loyal Customers
    (rfm["R_score"] >= 3) &
    (rfm["F_score"] >= 4) &
    (rfm["M_score"] >= 3),

    # 3. Potential Loyalists
    (rfm["R_score"] >= 4) &
    (rfm["F_score"] <= 3) &
    (rfm["M_score"] >= 3),

    # 4. High-Value At Risk
    (rfm["R_score"] <= 2) &
    (rfm["F_score"] >= 4) &
    (rfm["M_score"] >= 4),

    # 5. At Risk
    (rfm["R_score"] <= 2) &
    (
        (rfm["F_score"] >= 3) |
        (rfm["M_score"] >= 3)
    ),

    # 6. Hibernating
    (rfm["R_score"] <= 2) &
    (rfm["F_score"] <= 2) &
    (rfm["M_score"] <= 2),

    # 7. New / Promising
    (rfm["R_score"] >= 4) &
    (rfm["F_score"] <= 3) &
    (rfm["M_score"] <= 2)
]

segments = [
    "Champions",
    "Loyal Customers",
    "Potential Loyalists",
    "High-Value At Risk",
    "At Risk",
    "Hibernating",
    "New / Promising"
]
```

```python
rfm["customer_segment_rfm"] = np.select(
    conditions,
    segments,
    default="Regular Customers"
)
```

```python
ml_features = [
    "age",
    "active_products",
    "app_logins_frequency",
    "feature_usage_diversity",
    "international_transactions",
    "failed_transactions",
    "satisfaction_score",
    "nps_score",
    "support_tickets_count",
    "resolved_tickets_ratio",
    "app_store_rating",
    "monthly_transaction_count",
    "transaction_frequency",
    "weekend_transaction_ratio",
    "avg_daily_transactions",
    "customer_tenure",
    "customer_lifetime_value"
]

X = customers_clean[ml_features].copy()
y = customers_clean["churn_probability"].copy()

print("X shape:", X.shape)
print("Missing values:", X.isna().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

# Create the model
baseline_model = LinearRegression()

# Train
baseline_model.fit(X_train, y_train)

# Predict on unseen test data
y_pred = baseline_model.predict(X_test)

# Evaluate
mae = mean_absolute_error(y_test, y_pred)
rmse = np.sqrt(mean_squared_error(y_test, y_pred))
r2 = r2_score(y_test, y_pred)

print("=== BASELINE LINEAR REGRESSION ===")
print(f"MAE:  {mae:.4f}")
print(f"RMSE: {rmse:.4f}")
print(f"R²:   {r2:.4f}")
```

```python
transactions_clean = transactions_clean.sort_values(
    ["customer_id", "date"]
)

transactions_clean["days_since_previous"] = (
    transactions_clean
    .groupby("customer_id")["date"]
    .diff()
    .dt.days
)

print("=== TRANSACTION GAP DISTRIBUTION ===")

print(
    transactions_clean["days_since_previous"]
    .describe(
        percentiles=[0.50, 0.75, 0.90, 0.95, 0.99]
    )
)
```

```python
# Define the observation cutoff
observation_end = transactions_clean["date"].max()

# 30-day inactivity threshold
churn_threshold = 30

# Last transaction for each customer
last_transaction = (
    transactions_clean
    .groupby("customer_id")["date"]
    .max()
)

# Days inactive as of the dataset end
customer_inactivity = (
    observation_end - last_transaction
).dt.days

print("Observation end:", observation_end)
print("Churn threshold:", churn_threshold, "days")

print("\n=== INACTIVITY DISTRIBUTION ===")
print(customer_inactivity.describe())

print("\n=== POTENTIAL CHURN LABEL ===")
print(
    (customer_inactivity >= churn_threshold)
    .value_counts()
)
```

```python
# Create actual behavioral churn label
churn_label = (
    customer_inactivity >= churn_threshold
).astype(int)

# Create customer-level churn dataframe
churn_data = pd.DataFrame({
    "customer_id": customer_inactivity.index,
    "days_inactive": customer_inactivity.values,
    "churned": churn_label.values
})

print("=== ACTUAL CHURN LABEL ===")
print(churn_data["churned"].value_counts())

print("\nChurn rate:")
print(churn_data["churned"].mean())
```

```python
churn_model_data = customers_clean.merge(
    churn_data,
    on="customer_id",
    how="inner"
)

print("Churn modeling dataset:", churn_model_data.shape)

print("\nMissing churn labels:",
      churn_model_data["churned"].isna().sum())

print("\nChurn distribution:")
print(
    churn_model_data["churned"]
    .value_counts()
)
```

```python
# Features we should NOT use
exclude_columns = [
    "customer_id",
    "churned",
    "days_inactive",
    "churn_probability"
]

# Create feature matrix
X_churn = churn_model_data.drop(
    columns=exclude_columns,
    errors="ignore"
)

y_churn = churn_model_data["churned"]

print("X shape:", X_churn.shape)
print("y shape:", y_churn.shape)

print("\nData types:")
print(X_churn.dtypes.value_counts())

print("\nMissing values:")
print(X_churn.isna().sum().sum())
```

```python
churn_features = [
    "age",
    "household_size",
    "income_bracket",
    "active_products",
    "app_logins_frequency",
    "feature_usage_diversity",
    "credit_utilization_ratio",
    "international_transactions",
    "failed_transactions",
    "tx_count",
    "avg_tx_value",
    "total_tx_volume",
    "base_satisfaction",
    "tx_satisfaction",
    "product_satisfaction",
    "satisfaction_score",
    "nps_score",
    "support_tickets_count",
    "resolved_tickets_ratio",
    "app_store_rating",
    "monthly_transaction_count",
    "average_transaction_value",
    "total_transaction_volume",
    "transaction_frequency",
    "weekend_transaction_ratio",
    "avg_daily_transactions",
    "customer_tenure",
    "customer_lifetime_value"
]

X_churn = churn_model_data[churn_features].copy()
y_churn = churn_model_data["churned"]

print("X shape:", X_churn.shape)
print("Missing values:", X_churn.isna().sum().sum())
```

```python
from sklearn.impute import SimpleImputer

# Separate numeric and categorical columns
numeric_features = X_churn.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_features = X_churn.select_dtypes(
    include=["object", "category", "bool"]
).columns

print("Numeric features:", len(numeric_features))
print("Categorical features:", len(categorical_features))
print("\nCategorical columns:")
print(list(categorical_features))
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# Numeric columns
numeric_imputer = SimpleImputer(strategy="median")

X_numeric = pd.DataFrame(
    numeric_imputer.fit_transform(X_churn[numeric_features]),
    columns=numeric_features,
    index=X_churn.index
)

# Categorical column
categorical_imputer = SimpleImputer(strategy="most_frequent")

X_categorical = pd.DataFrame(
    categorical_imputer.fit_transform(X_churn[categorical_features]),
    columns=categorical_features,
    index=X_churn.index
)

print("Numeric missing:", X_numeric.isna().sum().sum())
print("Categorical missing:", X_categorical.isna().sum().sum())
```

```python
encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

income_encoded = pd.DataFrame(
    encoder.fit_transform(X_categorical),
    columns=encoder.get_feature_names_out(categorical_features),
    index=X_churn.index
)

print("Encoded columns:")
print(list(income_encoded.columns))
```

```python
X_churn_processed = pd.concat(
    [X_numeric, income_encoded],
    axis=1
)

print("Final X shape:", X_churn_processed.shape)
print("Missing values:", X_churn_processed.isna().sum().sum())
```

```python
from sklearn.model_selection import train_test_split

X_train_churn, X_test_churn, y_train_churn, y_test_churn = train_test_split(
    X_churn_processed,
    y_churn,
    test_size=0.20,
    random_state=42,
    stratify=y_churn
)

print("Training:", X_train_churn.shape)
print("Testing:", X_test_churn.shape)

print("\nTraining churn rate:", y_train_churn.mean())
print("Testing churn rate:", y_test_churn.mean())
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

logistic_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

logistic_model.fit(
    X_train_churn,
    y_train_churn
)

logistic_pred = logistic_model.predict(X_test_churn)
logistic_prob = logistic_model.predict_proba(X_test_churn)[:, 1]

print("=== LOGISTIC REGRESSION ===")
print(f"Accuracy:  {accuracy_score(y_test_churn, logistic_pred):.4f}")
print(f"Precision: {precision_score(y_test_churn, logistic_pred):.4f}")
print(f"Recall:    {recall_score(y_test_churn, logistic_pred):.4f}")
print(f"F1 Score:  {f1_score(y_test_churn, logistic_pred):.4f}")
print(f"ROC-AUC:   {roc_auc_score(y_test_churn, logistic_prob):.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_churn, logistic_pred))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_churn = RandomForestClassifier(
    n_estimators=300,
    random_state=42,
    n_jobs=-1,
    class_weight="balanced"
)

rf_churn.fit(
    X_train_churn,
    y_train_churn
)

rf_churn_pred = rf_churn.predict(X_test_churn)
rf_churn_prob = rf_churn.predict_proba(X_test_churn)[:, 1]

print("=== RANDOM FOREST CHURN CLASSIFIER ===")
print(f"Accuracy:  {accuracy_score(y_test_churn, rf_churn_pred):.4f}")
print(f"Precision: {precision_score(y_test_churn, rf_churn_pred):.4f}")
print(f"Recall:    {recall_score(y_test_churn, rf_churn_pred):.4f}")
print(f"F1 Score:  {f1_score(y_test_churn, rf_churn_pred):.4f}")
print(f"ROC-AUC:   {roc_auc_score(y_test_churn, rf_churn_prob):.4f}")

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_churn, rf_churn_pred))
```

```python
import numpy as np

threshold_results = []

for threshold in np.arange(0.20, 0.61, 0.05):

    pred = (rf_churn_prob >= threshold).astype(int)

    threshold_results.append({
        "threshold": threshold,
        "precision": precision_score(y_test_churn, pred),
        "recall": recall_score(y_test_churn, pred),
        "f1": f1_score(y_test_churn, pred),
        "accuracy": accuracy_score(y_test_churn, pred)
    })

threshold_results = pd.DataFrame(threshold_results)

print(threshold_results.round(4).to_string(index=False))
```

```python
final_threshold = 0.40

final_pred = (
    rf_churn_prob >= final_threshold
).astype(int)

final_cm = confusion_matrix(
    y_test_churn,
    final_pred
)

tn, fp, fn, tp = final_cm.ravel()

print("=== FINAL CHURN MODEL @ 0.40 ===")

print("True Negatives:", tn)
print("False Positives:", fp)
print("False Negatives:", fn)
print("True Positives:", tp)

print("\nCustomers flagged as churn risk:", tp + fp)
print("Actual churners caught:", tp)
print("Actual churners:", tp + fn)

print("\nRecall:", recall_score(y_test_churn, final_pred))
print("Precision:", precision_score(y_test_churn, final_pred))
print("F1:", f1_score(y_test_churn, final_pred))
print("ROC-AUC:", roc_auc_score(y_test_churn, rf_churn_prob))
```
